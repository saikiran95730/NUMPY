```python
import matplotlib.pyplot as plt 
import numpy as np 

x = np.linspace(0, 10, 100) 

# Create a 2-row by 1-column grid, select panel 1 
plt.subplot(2, 1, 1) 
plt.plot(x, np.sin(x), 'b-') 
plt.title('Top Subplot: Sine') 

# Select panel 2 
plt.subplot(2, 1, 2) 
plt.plot(x, np.cos(x), 'r--') 
plt.title('Bottom Subplot: Cosine') 
plt.show()
```

```python
import matplotlib.pyplot as plt 
categories = ['Engineering', 'Business', 'Arts', 'Science'] 
enrollment = [45, 30, 15, 25] 

# Vertical Bar Chart 
plt.bar(categories, enrollment, color='teal') 
plt.xlabel('Department') 
plt.ylabel('Students Enrolled') 
plt.title('Department Enrollment Comparison') 
plt.show()
```

```python
import matplotlib.pyplot as plt 
categories = ['Q1', 'Q2', 'Q3', 'Q4'] 
group_a = [15, 30, 45, 22] 
group_b = [15, 25, 50, 20] 

plt.bar(categories, group_a, color='blue', label='Group A') 
plt.bar(categories, group_b, color='red', bottom=group_a, label='Group B') 
plt.ylabel('Total Units') 
plt.title('Stacked Quarterly Sales') 
plt.legend() 
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 4, 0.3)
y = np.exp(-x)
error = 0.1 * np.abs(np.random.randn(len(y)))

plt.errorbar(x, y, yerr=error, fmt='.-', color='navy')
plt.xlabel('Time Step')
plt.ylabel('Decay Measurement')
plt.title('Experimental Decay with Error Bars')
plt.show()
```
