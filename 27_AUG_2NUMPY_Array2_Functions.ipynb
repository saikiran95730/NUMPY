# 2. Array Manipulating functions:

```python
a1 = np.array([1,2,3])
reshaped = np.reshape(a1, (1,3)) # Reshape 1x3
print("Reshaped Array:", reshaped)
```

```python
a2 = np.array([1,2,3])
reshaped = np.reshape(a2, (3,1)) # Reshape 3x1
print("Reshaped Array:\n", reshaped)
```

```python
f1 = np.array([[1,3], [5,6], [7,4], [2,8]])
flattened = np.ravel(f1,) # Flatten to 1D array
print("Flattened Array:", flattened)
```

```python
f1 = np.array([[1,3], [5,6], [7,4], [2,8]])
print("Flattened Array:", f1.flatten())
```

```python
e1 = np.array([[1,2],[3,4]])
transposed = np.transpose(e1) # Transpose the array
print("Transposed Array:\n", transposed)
```

```python
In NumPy, transpose converts rows into columns and columns into rows.
    By interchanging the values.
```

```python
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.vstack([a2,b2]) # Stack a2 abd b2 vertically
print("Vertically Stacked Array:\n", stacked)
```

```python
a3 = np.array([1,2])
b3 = np.array([3,4])
stacked1 = np.hstack([a3,b3]) # Stack a2 abd b2 horizontally
print("Horizontally Stacked Array:\n", stacked1)
```

# 4. Statistical Functions

```python
s = np.array([1,2,3,4])
mean = np.mean(s)
print("Mean of s:", mean)
```

```python
variance = np.var(s)
print("Variance of s:", variance)
```

```python
std_devi = np.std(s)
print("Standard deviation of s:", std_devi)  ## √(1.25)
```

```python
minimum = np.min(s)
print("Minimum:", minimum)
```

```python
maximum = np.max(s)
print("Maximum:", maximum)
```

# 7. Boolean & Logical functions

```python
# Check if all elements are True

# all

logical_test = np.array([True, False, True])
all_true = np.all(logical_test) # Check if all are True
print ("All elements in array are True:", all_true)
```

```python
# Check if any elements are True

# any

logical_test = np.array([True, False, True])
any_true = np.any(logical_test) # Check if all are True
print ("Any one elements atleast in array are True:", any_true)
```

# 8. Set Operations

```python
import numpy as np
set_a = np.array([1, 2, 3, 4])
set_b = np.array([3, 4, 5, 6])
```

```python
# Union of two arrays

union = np.union1d(set_a, set_b)
print ("Union of a and b:", union)
```

```python
# Intersection of two arrays

intersection = np.intersect1d(set_a, set_b)
print ("Intersection of a and b:", intersection)
 
```

# 9. Array Attributes functions

```python
# Array attributes
import numpy as np

a = np.array([1, 2, 3])

shape = a.shape # Shape of the array
size = a.size # Number of elements
dimensions = a.ndim # Number of dimensions
dtype = a.dtype # Data type of the array.

print("Shape of a:", shape)
print("Size of a:", size)
print("Number of dimensions of a:", dimensions)
print("Data type of a:", dtype)
```

# 10. Other functions

```python
# Create a copy of an array
import numpy as np 

a = np.array([1,2,3])
copied_array_a = np.copy(a) # Create a copy of array a
print ("Copied array:", copied_array_a)
```

```python
# Size in bytes of an array

array_size_in_bytes = a.nbytes # Size in bytes
print("Size of a in bytes:", array_size_in_bytes)
```

```python
# Check if two arrays share memory
shared = np.shares_memory(a, copied_array_a)
print ("Do a and copied_array_a  share memory?", shared) # Check if arrays share memory
```
