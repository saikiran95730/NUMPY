## Python program to generate OTP

```python
import random

def generate_otp(length):
    digits = '01234'
    otp = ''.join(random.choice(digits) for _ in range(length))
    return otp

otp_length = 5
otp = generate_otp(otp_length)
print(f'your otp is {otp}')
```

```python
# importing Numpy package 
import numpy as np

# creating a Numpy array
num_array = np.arange(100).reshape(20, 5)
print(num_array)


# Display array in C order

print("\n  C order array:")
for num_array in np.nditer(num_array, order="C"):
	print(num_array,end=' ')
```

```python
# importing Numpy package 
import numpy as np

# creating a Numpy array
num_array = np.arange(100).reshape(20, 5)
print(num_array)


# Display array in Fortran order

print("\n  Fortran order array:")
for num_array in np.nditer(num_array, order="F"):
	print(num_array,end=' ')
```
